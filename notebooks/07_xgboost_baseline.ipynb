# 07 XGBoost Baseline
Train the simple baseline XGBoost configuration specified in the brief.

```python
from pathlib import Path
import sys
import joblib
import pandas as pd
from sklearn.pipeline import Pipeline

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))
from src.config import MODEL_DIR, PROCESSED_DATA_DIR
from src.preprocessing import make_xgboost_preprocessor
from src.xgboost_model import baseline_xgboost

X_train = pd.read_csv(PROCESSED_DATA_DIR / "X_train.csv")
y_train = pd.read_csv(PROCESSED_DATA_DIR / "y_train.csv").squeeze("columns")
X_test = pd.read_csv(PROCESSED_DATA_DIR / "X_test.csv")
y_test = pd.read_csv(PROCESSED_DATA_DIR / "y_test.csv").squeeze("columns")
model = Pipeline([("preprocess", make_xgboost_preprocessor(X_train)), ("model", baseline_xgboost())])
model.fit(X_train, y_train)
probabilities = model.predict_proba(X_test)[:, 1]
joblib.dump(model, MODEL_DIR / "baseline_xgboost" / "notebook_model.joblib")
print("Baseline model trained. Test rows:", len(y_test))
probabilities[:5]
```

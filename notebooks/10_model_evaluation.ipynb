# 10 Model Evaluation
Compare baseline XGBoost, tuned XGBoost, and tuned CatBoost using ROC-AUC, PR-AUC, F1, recall, precision, calibration, and business cost.

```python
from pathlib import Path
import sys
import joblib
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import RocCurveDisplay, PrecisionRecallDisplay

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))
from src.config import MODEL_DIR, PROCESSED_DATA_DIR
from src.evaluation import evaluate_binary

X_test = pd.read_csv(PROCESSED_DATA_DIR / "X_test.csv")
y_test = pd.read_csv(PROCESSED_DATA_DIR / "y_test.csv").squeeze("columns")
model_path = MODEL_DIR / "baseline_xgboost" / "notebook_model.joblib"
if not model_path.exists():
    model_path = MODEL_DIR / "baseline_xgboost" / "model.joblib"
model = joblib.load(model_path)
probabilities = model.predict_proba(X_test)[:, 1]
metrics = evaluate_binary(y_test, probabilities)
print(pd.Series(metrics))
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
RocCurveDisplay.from_predictions(y_test, probabilities, ax=axes[0])
PrecisionRecallDisplay.from_predictions(y_test, probabilities, ax=axes[1])
plt.tight_layout()
plt.show()
```
